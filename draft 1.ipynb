```python
import csv

with open("spending_data.csv", newline="", encoding="utf-8") as f: #didn't wrie with w because w means wrie and erase
    reader = csv.reader(f)
    header = next(reader)      #reads the first row
    data = list(reader)    #read the all other rows as a big data   

print("Number of rows:", len(data))
print("Number of columns:", len(header))


seen_rows = []
duplicate_rows = 0
for row in data:
    if row in seen_rows:
        duplicate_rows = duplicate_rows + 1
    else:
        seen_rows.append(row)
print("Exact duplicate rows:", duplicate_rows)


seen_ids = []
repeated_ids = []
for row in data:
    customer_id = row[0]       
    if customer_id in seen_ids:
        repeated_ids.append(customer_id)
    else:
        seen_ids.append(customer_id)
print("Repeated customer IDs:", len(repeated_ids))
print(repeated_ids)
```
